```python
# EXPERIMENT 4: DATA EXPLORATION (EDA)
# Heart Disease Prediction

# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# STEP 1: LOAD THE DATASET

# Load the original dataset
df = pd.read_csv("Heart_disease_cleveland_new.csv")

print("Dataset shape:", df.shape)

# Display first 5 rows
df.head()
```

```python
# STEP 2: BASIC DATA CHECK

# Check column names
print("Columns:")
print(df.columns.tolist())

# Check data types
print("\nData types:")
print(df.dtypes)

# Check missing values
print("\nMissing values:")
print(df.isnull().sum())
```

```python
# STEP 3: TARGET DISTRIBUTION

# Count the two target classes
print(df["target"].value_counts())

# Simple bar graph
df["target"].value_counts().sort_index().plot(kind="bar")

plt.title("Heart Disease Distribution")
plt.xlabel("Target (0 = No Disease, 1 = Disease)")
plt.ylabel("Number of Patients")
plt.show()
```

```python
# STEP 4: DISTRIBUTION OF NUMERICAL FEATURES

# Numerical features
numerical_features = [
    "age",
    "trestbps",
    "chol",
    "thalach",
    "oldpeak"
]

# Draw a histogram for each numerical feature
for column in numerical_features:
    plt.figure(figsize=(5, 3))
    plt.hist(df[column], bins=15, edgecolor="black")
    plt.title("Distribution of " + column)
    plt.xlabel(column)
    plt.ylabel("Frequency")
    plt.show()
```

```python
# STEP 5: DISTRIBUTION OF CATEGORICAL FEATURES

# Categorical features
categorical_features = [
    "sex",
    "cp",
    "fbs",
    "restecg",
    "exang",
    "slope",
    "ca",
    "thal"
]

# Draw a simple bar graph for each categorical feature
for column in categorical_features:
    df[column].value_counts().sort_index().plot(kind="bar")

    plt.title("Distribution of " + column)
    plt.xlabel(column)
    plt.ylabel("Number of Patients")
    plt.show()
```

```python
# STEP 6: CHEST PAIN TYPE VS HEART DISEASE

# Count target values for each chest pain type
cp_target = pd.crosstab(df["cp"], df["target"])

print(cp_target)

# Simple grouped bar graph
cp_target.plot(kind="bar")

plt.title("Chest Pain Type vs Heart Disease")
plt.xlabel("Chest Pain Type")
plt.ylabel("Number of Patients")
plt.legend(["No Disease", "Disease"])
plt.show()
```

```python
# STEP 7: MAXIMUM HEART RATE VS HEART DISEASE

# Compare maximum heart rate for both target classes
plt.boxplot([
    df[df["target"] == 0]["thalach"],
    df[df["target"] == 1]["thalach"]
])

plt.xticks([1, 2], ["No Disease", "Disease"])
plt.title("Maximum Heart Rate vs Heart Disease")
plt.ylabel("Maximum Heart Rate")
plt.show()

# Mean maximum heart rate for each target class
print(df.groupby("target")["thalach"].mean())
```

```python
# STEP 8: OLDPEAK VS HEART DISEASE

# Compare oldpeak for both target classes
plt.boxplot([
    df[df["target"] == 0]["oldpeak"],
    df[df["target"] == 1]["oldpeak"]
])

plt.xticks([1, 2], ["No Disease", "Disease"])
plt.title("Oldpeak vs Heart Disease")
plt.ylabel("Oldpeak")
plt.show()

# Mean oldpeak for each target class
print(df.groupby("target")["oldpeak"].mean())
```

```python
# STEP 9: CORRELATION ANALYSIS

# Calculate correlation between numerical features and target
corr = df.corr(numeric_only=True)

print(corr)

# Simple correlation heatmap
plt.figure(figsize=(9, 7))
sns.heatmap(corr, annot=True, cmap="coolwarm")

plt.title("Correlation Matrix")
plt.show()
```

```python
# STEP 10: OUTLIER CHECK

# Use boxplots to visually check numerical features for outliers
for column in numerical_features:
    plt.figure(figsize=(5, 3))
    plt.boxplot(df[column])
    plt.title("Outlier Check - " + column)
    plt.ylabel(column)
    plt.show()
```

```python
# STEP 11: COMPARE NUMERICAL FEATURES BY TARGET

# Find the mean of numerical features for each target class
mean_values = df.groupby("target")[numerical_features].mean()

print(mean_values)
```
